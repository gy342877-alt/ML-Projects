```python
data["mileage"] = data["mileage"].astype(str).str.extract(r"([\d.]+)", expand=False).astype(float)

data["engine"] = data["engine"].astype(str).str.extract(r"([\d.]+)", expand=False).astype(float)

data["max_power"] = data["max_power"].astype(str).str.extract(r"([\d.]+)", expand=False).astype(float)
```

```python
#data = data.drop("torque", axis=1) for deleteing any features 
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    data,
    test_size=0.2,
    random_state=42
)
```

```python
train_data.info()
```

```python
test_data.info()
```

```python
X_train = train_data.drop("selling_price", axis=1)
y_train = train_data["selling_price"]

X_test = test_data.drop("selling_price", axis=1)
y_test = test_data["selling_price"]
```

```python
num_cols = [
    "year",
    "km_driven",
    "mileage",
    "engine",
    "max_power",
    "seats"
]

cat_cols = [
    "fuel",
    "seller_type",
    "transmission",
    "owner"
]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
#combining these two pipeline 
preprocessing = ColumnTransformer([
    ("num", num_pipeline, num_cols),
    ("cat", cat_pipeline, cat_cols)
])
```

```python
X_train_prepared = preprocessing.fit_transform(X_train)
X_test_prepared = preprocessing.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
```

```python
lin_reg.fit(X_train_prepared, y_train)

```

```python
test_preds = lin_reg.predict(X_test_prepared)
```

```python
from sklearn.metrics import mean_squared_error

test_rmse = mean_squared_error(y_test, test_preds) ** 0.5

print("Linear Regression Test RMSE:", test_rmse)
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(random_state=42)

tree_reg.fit(X_train_prepared, y_train)

tree_preds = tree_reg.predict(X_test_prepared)

tree_rmse = mean_squared_error(y_test, tree_preds) ** 0.5

print("Decision Tree Test RMSE:", tree_rmse)
```

```python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(random_state=42)

tree_reg.fit(X_train_prepared, y_train)

tree_preds = tree_reg.predict(X_test_prepared)

tree_rmse = mean_squared_error(y_test, tree_preds) ** 0.5

print("Decision Tree Test RMSE:", tree_rmse)
```

```python
from sklearn.ensemble import RandomForestRegressor
forest_reg = RandomForestRegressor(random_state=42)

forest_reg.fit(X_train_prepared, y_train)

tree_preds = forest_reg.predict(X_test_prepared)

forest_reg = mean_squared_error(y_test, tree_preds) ** 0.5

print("RandomForestRegressor:",forest_reg)

```

```python
from sklearn.linear_model import Ridge
ridge_reg = Ridge()

ridge_reg.fit(X_train_prepared, y_train)

ridge_preds = ridge_reg.predict(X_test_prepared)

ridge_rmse = mean_squared_error(y_test, ridge_preds) ** 0.5

print("Ridge Test RMSE:", ridge_rmse)
```

```python
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error

svm_reg = SVR(kernel="rbf")

svm_reg.fit(X_train_prepared, y_train)

svm_preds = svm_reg.predict(X_test_prepared)

svm_rmse = mean_squared_error(y_test, svm_preds) ** 0.5

print("SVR Test RMSE:", svm_rmse)
```
